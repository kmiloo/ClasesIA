```python
# Paso 1: Importar bibliotecas necesarias
import pandas as pd
import numpy as np

# Paso 2: Definir los nombres de columnas originales
columnas_originales = [
    'A¥O', 'MES', '??Femicidio', '??Femicidio no ¡ntimo',
    '??Suicidio femicida', '??Violaci¢n con homicidio',
    '??Robo con homicidio', '??Auxilio al suicidio', '??Otros homicidios',
    '??Tortura o apremios ileg¡timos con violaci¢n', '??Robo con violaci¢n',
    '??Otras violaciones', '??Abusos sexuales', '??Otros delitos sexuales',
    '??Robos con violencia o intimidaci¢n',
    '??Robo violento de veh¡culo motorizado',
    '??Violencia intrafamiliar a mujer',
    '??Violencia intrafamiliar a ni¤o o ni¤a',
    '??Violencia intrafamiliar a adulto mayor',
    '??Violencia intrafamiliar a hombre',
    '??Violencia intrafamiliar no clasificada', '??Tr fico de sustancias',
    '??Microtr fico de sustancias',
    '??Elaboraci¢n o producci¢n de sustancias',
    '??Otras infracciones a la ley de drogas', '??Disparo injustificado',
    '??Porte / posesi¢n de armas o explosivos',
    '??Otras infracciones a la ley de armas', '??Robo en lugar habitado',
    '??Robos en lugar no habitado', '??Robo de veh¡culo motorizado',
    '??Robo de objetos de o desde veh¡culo',
    '??Animales sueltos en la v¡a p£blica', '??Comercio ilegal',
    '??Ofensas al pudor', '??Otras incivilidades', '??Abigeato',
    '??Robo frustrado', '??Otros delitos o faltas'
]

# Paso 3: Función para limpiar nombres de columnas
def limpiar_nombres_columnas(columnas):
    # Diccionario de reemplazos para caracteres especiales
    reemplazos = {
        '¥': 'Ñ',
        '¢': 'ó',
        '¡': 'í',
        '¤': 'ñ',
        '£': 'ú',
        '¦': 'á',
        '§': 'é',
        '¨': 'í',
        '©': 'ó',
        'ª': 'ú',
        '«': 'ü',
        '¬': 'ñ',
        '­': 'Ñ'
    }
    
    columnas_limpias = []
    for col in columnas:
        # Eliminar prefijo '??'
        if col.startswith('??'):
            col = col[2:]
        
        # Reemplazar caracteres especiales
        for char_erroneo, char_correcto in reemplazos.items():
            col = col.replace(char_erroneo, char_correcto)
        
        # Reemplazos específicos para palabras completas
        reemplazos_especificos = {
            'Tr fico': 'Tráfico',
            'Microtr fico': 'Microtráfico',
            'v¡a': 'vía',
            'p£blica': 'pública',
            'veh¡culo': 'vehículo',
            'ni¤o': 'niño',
            'ni¤a': 'niña',
            'intimidaci¢n': 'intimidación',
            'ileg¡timos': 'ilegítimos',
            'posesi¢n': 'posesión',
            'Elaboraci¢n': 'Elaboración',
            'producci¢n': 'producción',
            'violaci¢n': 'violación'
        }
        
        for palabra_erronea, palabra_correcta in reemplazos_especificos.items():
            col = col.replace(palabra_erronea, palabra_correcta)
        
        columnas_limpias.append(col)
    
    return columnas_limpias

# Paso 4: Aplicar la limpieza
columnas_limpias = limpiar_nombres_columnas(columnas_originales)

# Paso 5: Mostrar resultados
print("Nombres de columnas originales vs limpios:")
print("-" * 60)
for orig, limp in zip(columnas_originales, columnas_limpias):
    print(f"Original: {orig}\nLimpio:   {limp}")
    print("-" * 60)

# Paso 6: Crear un DataFrame de ejemplo para demostración
# (En tu caso real, usarías tus datos reales aquí)
data_ejemplo = {col: [] for col in columnas_limpias}
df = pd.DataFrame(data_ejemplo)

# Mostrar el DataFrame con los nombres limpios
print("\nDataFrame con columnas limpias:")
df
```
